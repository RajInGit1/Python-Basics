Question:
Given an array of integers, find the length of the longest consecutive sequence of numbers.

Example:

Input:  [100, 4, 200, 1, 3, 2]
Output: 4

Explanation: The longest consecutive sequence is 1, 2, 3, 4, whose length is 4.

def longest_consecutive(nums):
    num_set = set(nums)
    longest = 0

    for num in num_set:
        # Start counting only if num is the beginning
        # of a consecutive sequence.
        if num - 1 not in num_set:
            current = num
            length = 1

            while current + 1 in num_set:
                current += 1
                length += 1

            longest = max(longest, length)

    return longest


nums = [100, 4, 200, 1, 3, 2]

print(longest_consecutive(nums))
