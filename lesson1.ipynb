```python
import numpy as np
arr = np.array([1,2,3,4,5,6], dtype=float)
print(arr)
```

```python
arr = np.array([[1,2,3],[1,2,3]])
print(arr)
```

```python
arr = np.zeros([5,5])
print(arr)
```

```python
arr = np.ones([5,5])
print(arr)
```

```python
arr = np.empty(1)
print(arr)
```

```python
arr = np.full([2,3],7)
print(arr)
```

```python
arr = np.eye(7)
print(arr)
```

```python
arr = np.identity(7)
print(arr)
```

```python
arr = np.arange(0,10,2)
print(arr)
```

```python
arr = np.linspace(0,10,5)
print(arr)
```

```python
arr = np.logspace(0,10,3)
print(arr)
```

```python
import pandas as pd

data = {
    "Name": ["Alice", "Bob", "Charlie"],
    "Marks": [85, 92, 78],
    "Grade": ["B", "A", "C"]
}

df = pd.DataFrame(data)

df
```

```python
df.style \
    .set_caption("Student Marks") \
    .background_gradient(cmap="Blues") \
    .set_properties(**{
        "text-align": "center",
        "font-size": "14px"
    })
```

```python
df.style \
    .set_caption("Student Marks") \
    .set_table_styles([
        {
            "selector": "th",
            "props": [
                ("background-color", "#2E86C1"),
                ("color", "white"),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        }
    ])
```

```python
df.style \
    .highlight_max(subset=["Marks"], color="lightgreen")
```

```python
df.style \
    .set_caption("📊 Student Performance") \
    .set_table_styles([
        {
            "selector": "caption",
            "props": [
                ("font-size", "20px"),
                ("font-weight", "bold"),
                ("color", "#2E86C1")
            ]
        },
        {
            "selector": "th",
            "props": [
                ("background-color", "#2E86C1"),
                ("color", "white"),
                ("font-weight", "bold"),
                ("text-align", "center")
            ]
        },
        {
            "selector": "td",
            "props": [
                ("text-align", "center"),
                ("padding", "8px")
            ]
        }
    ]) \
    .highlight_max(subset=["Marks"], color="lightgreen")
```
